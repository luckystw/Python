```python
path = r"C:\Users\hp\Desktop\GEN AI Assignments\GEN AI - Assignment 4 - Lucky Satyawali\Gen AI-Task(1)-Lucky Satyawali\sales_data.txt"

new_sales = [5000, 2500, 1700]

file = open(path, "r")
existing = file.read()
file.close()

file = open(path, "a")
if existing and not existing.endswith("\n"):
    file.write("\n")
for sale in new_sales:
    file.write(str(sale) + "\n")
file.close()

file = open(path, "r")
print(file.read())
file.close()

file = open(path, "r")
lines = file.readlines()
print("Total lines:", len(lines))
file.close()
```
