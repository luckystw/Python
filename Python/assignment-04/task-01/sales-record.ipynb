```python
sales = [1200, 450, 980, 1500, 3000]

file = open("sales_data.txt", "w")
for sale in sales:
    file.write(str(sale) + "\n")
file.close()

file = open("sales_data.txt", "r")
print(file.read())
file.close()

file = open("sales_data.txt", "w")
file.write(",".join(str(sale) for sale in sales))
file.close()

file = open("sales_data.txt", "r")
print(file.read())
file.close()
```
