```python
path = r"C:\Users\hp\Desktop\GEN AI Assignments\GEN AI - Assignment 4 - Lucky Satyawali\Gen AI-Task(1)-Lucky Satyawali\discount_report.txt"

prices = {
    "Mouse": 500,
    "Keyboard": 800,
    "Monitor": 7000,
    "Pendrive": 400,
    "Camera": 5000
}

discount = float(input("Enter discount percentage: "))

file = open(path, "w")
total_discounted = 0
count = 0

for product in prices:
    original = prices[product]
    discounted = original - (original * discount / 100)
    total_discounted += discounted
    count += 1
    file.write(product + " | " + str(original) + " | " + str(discounted) + "\n")

file.write("\nTotal Items: " + str(count) + "\n")
file.write("Average Discounted Price: " + str(total_discounted / count) + "\n")
file.close()

file = open(path, "r")
print(file.read())
file.close()
```
