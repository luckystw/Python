```python
path = r"C:\Users\hp\Desktop\GEN AI Assignments\GEN AI - Assignment 4 - Lucky Satyawali\Gen AI-Task(1)-Lucky Satyawali\sales_data.txt"

file = open(path, "r")
lines = file.readlines()
file.close()

sales = []
for line in lines:
    for value in line.strip().split(","):
        if value.strip():
            sales.append(int(value.strip()))

total = sum(sales)
highest = max(sales)
lowest = min(sales)
average = total / len(sales)

print("Total Sales:", total)
print("Highest Sale:", highest)
print("Lowest Sale:", lowest)
print("Average Sale:", average)
```
