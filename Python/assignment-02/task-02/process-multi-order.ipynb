```python
orders = [1200, 2500, 800, 1750, 3000]
total_revenue = 0
discounted = 0

print("Order\tDiscount%\tFinal")
print("-" * 28)

for order_amount in orders:
    if order_amount >= 2000:
        discount = 15
    elif order_amount >= 1500:
        discount = 10
    elif order_amount >= 1000:
        discount = 7
    else:
        discount = 0

    final_amount = order_amount * (100 - discount) / 100
    total_revenue += final_amount

    if discount > 0:
        discounted += 1

    print(order_amount, discount, final_amount, sep="\t\t")

print("-" * 28)
print("Total revenue after discounts:", total_revenue)
print("Orders with a discount:", discounted)
```
