```python
orders = []

while True:
    print("\n1 - Add order amount")
    print("2 - Show orders and totals")
    print("q - Quit")
    choice = input("Choose an option: ")

    if choice == "q":
        print("Goodbye!")
        break

    if choice == "1":
        amount = input("Enter order amount: ")
        if not amount.isdigit():
            print("Invalid amount, try again.")
            continue
        orders.append(int(amount))
        print("Order added.")

    elif choice == "2":
        if len(orders) == 0:
            print("No orders yet.")
            continue

        total = 0
        print("Order\tDiscount%\tFinal")
        for order_amount in orders:
            if order_amount >= 2000:
                discount = 15
            elif order_amount >= 1500:
                discount = 10
            elif order_amount >= 1000:
                discount = 7
            else:
                discount = 0

            final_amount = order_amount * (100 - discount) / 100
            total += final_amount
            print(order_amount, discount, final_amount, sep="\t\t")

        print("Total after discounts:", total)

    else:
        print("Invalid choice, try again.")
        continue
```
