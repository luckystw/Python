```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"C:\Users\hp\Desktop\GEN AI Assignments\GEN AI - Assignment 12 - Lucky Satyawali\Housing.csv")
df.head()
```

```python
g = sns.FacetGrid(df, col="furnishingstatus")
g.map_dataframe(sns.scatterplot, x="area", y="price")
plt.show()
```

```python
sns.relplot(data=df, x="area", y="price", hue="furnishingstatus", col="airconditioning")
plt.show()
```

```python
sns.catplot(data=df, x="furnishingstatus", y="price", kind="box", col="airconditioning")
plt.show()
```

```python
sns.displot(data=df, x="price", kde=True, col="furnishingstatus")
plt.show()
```
