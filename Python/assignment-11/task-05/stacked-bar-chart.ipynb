```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("C:\\Users\\hp\\Desktop\\GEN AI Assignments\\GEN AI - Assignment 11 - Lucky Satyawali\\Sales Dataset.csv")
```

```python
category_payment = df.pivot_table(index="Category", columns="PaymentMode", values="Amount", aggfunc="sum", fill_value=0)
```

```python
category_payment.plot(kind="bar", stacked=True, figsize=(12, 7))
plt.title("Total Sales by Category and Payment Mode")
plt.xlabel("Category")
plt.ylabel("Total Sales Amount")
plt.xticks(rotation=0)
plt.legend(title="Payment Mode")
plt.tight_layout()
plt.show()
```
