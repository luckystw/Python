```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("C:\\Users\\hp\\Desktop\\GEN AI Assignments\\GEN AI - Assignment 11 - Lucky Satyawali\\Sales Dataset.csv")
```

```python
category_share = df.groupby("Category")["Amount"].sum()
```

```python
plt.figure(figsize=(8, 8))
plt.pie(category_share.values, labels=category_share.index, autopct="%1.1f%%", startangle=90)
plt.title("Category Share of Total Sales")
plt.show()
```
