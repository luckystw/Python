```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("C:\\Users\\hp\\Desktop\\GEN AI Assignments\\GEN AI - Assignment 11 - Lucky Satyawali\\Sales Dataset.csv")
df["Order Date"] = pd.to_datetime(df["Order Date"])
```

```python
df["Year"] = df["Order Date"].dt.year
df["Month"] = df["Order Date"].dt.month
yearly_monthly = df.pivot_table(index="Month", columns="Year", values="Amount", aggfunc="sum", fill_value=0)
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]
years = yearly_monthly.columns
x = np.arange(len(yearly_monthly.index))
bar_width = 0.8 / len(years)
```

```python
plt.figure(figsize=(14, 7))
for i, year in enumerate(years):
    plt.bar(x + i * bar_width, yearly_monthly[year], width=bar_width, label=str(year))
plt.title("Monthly Sales Comparison Across Years")
plt.xlabel("Month")
plt.ylabel("Total Sales Amount")
plt.xticks(x + bar_width * (len(years) - 1) / 2, [months[m - 1] for m in yearly_monthly.index])
plt.legend(title="Year")
plt.grid(axis="y")
plt.show()
```
