```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("C:\\Users\\hp\\Desktop\\GEN AI Assignments\\GEN AI - Assignment 11 - Lucky Satyawali\\Sales Dataset.csv")
df["Order Date"] = pd.to_datetime(df["Order Date"])
```

```python
monthly_sales = df.groupby(df["Order Date"].dt.to_period("M"))["Amount"].sum()
monthly_sales.index = monthly_sales.index.to_timestamp()
```

```python
plt.figure(figsize=(12, 6))
plt.plot(monthly_sales.index, monthly_sales.values, marker="o", linewidth=2)
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales Amount")
plt.xticks(rotation=45)
plt.grid()
plt.show()
```
