```python
import numpy as np

sales = np.array([1200, 1500, 900, 2000, 1800, 1700, 1600])

print("Total weekly sales:", np.sum(sales))
print("Average daily sales:", np.mean(sales))
print("Highest sales day:", np.argmax(sales) + 1, "with", np.max(sales))
print("Lowest sales day:", np.argmin(sales) + 1, "with", np.min(sales))
print("Standard deviation:", np.std(sales))
print("Days above average:", np.where(sales > np.mean(sales))[0] + 1)
print("Sales on those days:", sales[sales > np.mean(sales)])
```
