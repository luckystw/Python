```python
import sqlite3
import pandas as pd
```

```python
conn = sqlite3.connect('sample.db')
cursor = conn.cursor()
cursor.execute('CREATE TABLE employees (id INTEGER PRIMARY KEY, name TEXT, department TEXT, salary REAL)')
conn.commit()
```
