```python
class Product:
    def __init__(self, name, price, category):
        self.name = name
        self.__price = price
        self.category = category

    def get_price(self):
        return self.__price

    def set_price(self, new_price):
        if new_price > 0:
            self.__price = new_price
        else:
            print("Invalid price. Price must be greater than 0.")

    def get_info(self):
        print(f"Name: {self.name}")
        print(f"Price: {self.__price}")
        print(f"Category: {self.category}")

    def apply_discount(self, percent):
        return self.__price - (self.__price * percent / 100)


class Laptop(Product):
    def __init__(self, name, price, category, ram, processor):
        super().__init__(name, price, category)
        self.ram = ram
        self.processor = processor

    def get_info(self):
        print("--- Laptop ---")
        print(f"Model: {self.name}")
        print(f"Price: Rs. {self.get_price()}")
        print(f"Category: {self.category}")
        print(f"RAM: {self.ram} GB")
        print(f"Processor: {self.processor}")


class Mobile(Product):
    def __init__(self, name, price, category, battery, camera_mp):
        super().__init__(name, price, category)
        self.battery = battery
        self.camera_mp = camera_mp

    def get_info(self):
        print("--- Mobile ---")
        print(f"{self.name} | {self.category}")
        print(f"Price: Rs. {self.get_price()}")
        print(f"Battery: {self.battery} mAh")
        print(f"Camera: {self.camera_mp} MP")


laptop1 = Laptop("Dell XPS", 85000, "Electronics", 16, "Intel i7")
laptop2 = Laptop("MacBook Air", 95000, "Electronics", 8, "Apple M2")
mobile1 = Mobile("iPhone 14", 70000, "Electronics", 3279, 12)
mobile2 = Mobile("Galaxy S23", 65000, "Electronics", 3900, 50)

products = [laptop1, mobile1, laptop2, mobile2]

for item in products:
    item.get_info()
    print()
```
