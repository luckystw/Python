```python
# Creating a set of categories from a list of products and their corresponding categories
products = ["Laptop", "Mouse", "Banana", "Apple", "T-Shirt", "Jeans", "Keyboard"]
categories = ["Electronics", "Electronics", "Fruit", "Fruit", "Clothing", "Clothing", "Electronics"]

# 1. Create a set of categories
categories_set = set(categories)
print("Categories set:", categories_set)

# 2. Add a new category, then try adding a duplicate
categories_set.add("Books")
print("After adding 'Books':", categories_set)

categories_set.add("Books")  
categories_set.add("Electronics")
print("After adding duplicates:", categories_set)

# 3. Check whether a category exists
print("Is 'Fruit' in the set?", "Fruit" in categories_set)
print("Is 'Toys' in the set?", "Toys" in categories_set)      

# Extra: total number of unique categories
print("Total unique categories:", len(categories_set))
```
