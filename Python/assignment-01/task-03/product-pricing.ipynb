```python
# 1. Create dictionary
price_dict = {"Laptop": 55000, "Mouse": 499.5, "Keyboard": 1299,
              "Banana": 60, "Apple": 180, "T-Shirt": 799.99}

# 2. Add, update, remove
price_dict["Headphones"] = 2499     
price_dict["Mouse"] = 549          

for name in ["Banana", "Tablet"]:  
    if name in price_dict:
        del price_dict[name]
    else:
        print(name, "not found !!!")

print(price_dict)

# 3. Average price
average = sum(price_dict.values()) / len(price_dict)
print("Average price:", round(average, 2))

# Extra: max and min price products
print("Max:", max(price_dict, key=price_dict.get))
print("Min:", min(price_dict, key=price_dict.get))
```
