```python
price_dict = {"Laptop": 55000, "Mouse": 549, "Keyboard": 1299,
              "Apple": 180, "T-Shirt": 799.99, "Headphones": 2499}
products = list(price_dict)
categories = ["Electronics", "Electronics", "Electronics",
              "Fruit", "Clothing", "Electronics"]

# 1. Catalog of (name, price, category)
catalog = [(products[i], price_dict[products[i]], categories[i])
           for i in range(len(products))]
print("Catalog:", catalog)

# 2. Category -> product names
category_to_products = {}
for name, price, cat in catalog:
    category_to_products.setdefault(cat, []).append(name)
print("Category map:", category_to_products)

# 3. Products in the biggest category
top = max(category_to_products, key=lambda c: len(category_to_products[c]))
print(top, category_to_products[top])
```
